```python
#Dictionary merging

p1 = {
    "name": "Eshan",
    "age": 22,
}

p2 = {
    "rollno": 17
}
p1.update(p2)

print(p1)
```

```python
#asc and des

p1 = {
    "name": "Eshan",
    "age": 22,
}

p2 = {
    "rollno": 17
}

p1.update(p2)

p3 = dict(sorted(p1.items()))
p4 = dict(sorted(p1.items(),reverse=True))

print(p3, p4)
```
