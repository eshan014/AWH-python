```python
#Question 4. Factorial

n = int(input("Enter number: "))
j=1
for i in range(1, n+1):
    j = j * i
print(j)
```

```python
#Question 6. Adding "ing" to the end of string and if present adding "ly"

word = input("Enter word: ")
if word[-3:] == "ing":
    word = word + "ly"
else:
    word = word + "ing"

print(word)
```

```python
#Leap year

year = int(input("Enter year: "))
leap = []
for i in range(2026, year+1):
    if i % 4 == 0:
        if i % 100 != 0:
            leap.append(i)
        else:
            if i % 400 == 0:
                leap.append(i)
print(leap)
        
```

```python
#List comprehension
#Question a

integers = [2, -4, 0, 42, -2, -6, 19]
newlist = [i for i in integers if i>0]
print(newlist)
```

```python
#List comprehension
#Question b

integers1 = [2, 4, 0, 42, 2, -6, 19]

newlist1 = [i**2 for i in integers1]

print(newlist1)
```

```python
#List comprehension
#Question c

word = "hefgwbwchhcfuwryiqp"
v = "aeiou"

newlist2 = [i for i in word if i in v]

print(newlist2)
```

```python
#List comprehension
#Question d

word = input("Enter word: ")

lists = [ord(i) for i in word]

print(lists)
```
