```python
list_1 = [1,2,3,4,5]
list_2 = [1,2,3,4,3]


#Question (a)
if len(list_1) == len(list_2):
    print("Same length")
else:
    print("Length is not the same")
    
    
#Question (b)
    
if sum(list_1) == sum(list_2):
    print(f"Same sum: {sum(list_1)}")
else:
    print("Sum is not the same")

    
list_3 = [i for i in list_2 if i in list_1]
if not list_3:
    print("No values occur in both")
else:
    print(f"Values occur in both: {list_3}")
    
```
